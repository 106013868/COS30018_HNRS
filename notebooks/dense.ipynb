```python
import sys
sys.path.append("..")

from hnrs.preprocessing import load_preprocessed_mnist

x_train, y_train, y_cat_train, x_test, y_test, y_cat_test = load_preprocessed_mnist()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
import keras
from keras import Input
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from sklearn.metrics import confusion_matrix
```

```python
#x_train = x_train.reshape(x_train.shape[0], -1)
#x_test = x_test.reshape(x_test.shape[0], -1)
#print(x_train.shape, x_test.shape)
```

```python
keras.utils.set_random_seed(42)
model = Sequential()

model.add(Input(shape=(28,28)))
model.add(Flatten())
model.add(Dense(units=128, activation='relu'))
model.add(Dense(units=64, activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(units=10, activation='softmax'))
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
model.summary()
```

```python
batch_size = 256
epochs = 10
model.fit(x=x_train, y=y_cat_train, batch_size=batch_size, epochs=epochs, validation_split=0.2)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_cat_test)
print("Test Loss: {}, Test Accuracy: {}".format(test_loss, test_acc))
```

```python
y_pred = model.predict(x_test).argmax(axis=1)
print(confusion_matrix(y_test, y_pred))
```

```python
model.save("../models/dense.keras")
```
